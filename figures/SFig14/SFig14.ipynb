# Supplementary Figure 14

Measured MS2 spectra for four selected PREDICT stool peptides with a histidine at a protein
boundary, from the non-specific Sage search. Each panel shows the processed spectrum (gray)
and the matched b (blue) and y (orange) ions. The notebook redraws the figure with the
original producer's code, removes the figure title as the release step did, re-derives each
case's processed peaks and matched ions from the measured peaks, and checks the case-selection
rule.

It also draws the v30 house-style version of the same values (`SFig14_v30`), the figure placed
from v30 on.

Run from the repository root: `python paper.py build SFig14`.

```python
from pathlib import Path
import sys
UNIT = Path.cwd()                 # the notebook runs with its own directory as cwd
REPO = UNIT.parents[1]
sys.path[:0] = [str(REPO), str(UNIT)]
SD = REPO / "source_data" / UNIT.name
OUT = REPO / "build" / UNIT.name   # rendered panels; gitignored
OUT.mkdir(parents=True, exist_ok=True)
from style import fl_style as fl
import panels
fl.init()

import os
import pandas as pd
pd.set_option("display.width", 200)
FONT_DIR = os.environ.get("FASTALAKE_FONT_DIR")
rederived = {}
```

## Panel a

Case HIS_001: human chymotrypsin C (CTRC) peptide SLFVGVDTIHVH, histidine at the C-terminal boundary. Supports the legend, v28d line 465: "Four selected PREDICT spectra from
the non-specific search." n = 1 spectrum (one PSM in one stool acquisition). No test;
the selection is illustrative.

Below, the processed peaks are re-derived from all observed peaks of the scan and the matched
b/y ions from the processed peaks, peptide, precursor charge and mass (port of the upstream
extraction code), and compared with the shipped tables and Sage's matched-peak count.

`panel_abcd` (pdf, svg, png at 180 dpi), and the title-removed composite as `SFig14_v28f`
(pdf, and png rasterised at 300 dpi).

```python
panels.figure_abcd(SD, OUT / "panel_abcd")
print(panels.remove_title(OUT / "panel_abcd.pdf", OUT / "SFig14_v28f.pdf", OUT / "SFig14_v28f.png"))
from IPython.display import Image, display
display(Image(filename=str(OUT / "SFig14_v28f.png"), width=720))
d = panels.stat_rederive_spectrum(SD, "HIS_001")
rederived["HIS_001"] = d
for k, v in d.items():
    print(k, "=", v)
assert d["processed_equal"] and d["annotations_equal"] and d["matched_ions"] == d["sage_matched_peaks"]
```

## v30 composite (house style)

The figure as placed in the manuscript from v30 on: the same data and plotted values, redrawn in
the house style (`panels.figure_v30`, port of the v30 restyle script). The function first runs the
producer drawing above with saving intercepted, then compares every plotted array of the restyled
figure with it exactly and raises on any difference. Output: `SFig14_v30.pdf` and the PDF
rasterised at 300 dpi, `SFig14_v30.png`. The house style needs the Söhne fonts
(`FASTALAKE_FONT_DIR`) for its text pixels.


```python
v30 = panels.figure_v30(SD, OUT / "SFig14_v30", OUT / "producer_reference")
for line in v30["panels"]:
    print(line)
print("labels displaced:", v30["labels_displaced"], "; clip check:", v30["clip_ok"])
display(Image(filename=str(OUT / "SFig14_v30.png"), width=720))
```

## Case selection and case table

The four cases were chosen by a rule fixed before the spectra were looked at (upstream
analysis A25): one case per origin (human, microbial) and boundary side (N, C) in PREDICT
stool; within each, the acquisition at the median candidate-count rank, then the PSM at the
median matched-intensity rank. The table checks that rule against the shipped selection
tables. The second table compares the case table with the upstream audit, the re-derivation
and the bounded-alternative tables, and re-derives its two flag columns: `accepted_in_semi`
(a same-scan semi-tryptic PSM passing both q gates; PREDICT had no semi-tryptic route, so
False) and `N_flanking_H_inferred` (side N). The producer's fragment ppm check is repeated last.

```python
sel = panels.stat_selection(SD)
print(sel.to_string(index=False))
assert sel.acquisition_rule_holds.all() and sel.psm_rule_holds.all()
cr = panels.stat_case_review(SD, list(rederived.values()))
print(cr.T.to_string())
assert (cr.accepted_in_semi_review == cr.accepted_in_semi_recomputed).all()
assert (cr.N_flanking_H_inferred_review == cr.N_flanking_H_inferred_recomputed).all()
print("fragment ppm check:", panels.check_fragment_ppm(SD))
```

## Checks and numbers

Pixel checks compare the regenerated figure with the original producer's PNG (180 dpi) and PDF
(300 dpi), and the title-removed composite with the image in the manuscript (PNG) and its PDF
(300 dpi). The producer used DejaVu Sans, which ships with matplotlib, so these checks do not
depend on the Söhne files; they run only when `FASTALAKE_FONT_DIR` is set, as for every unit.
The v30 house-style composite is compared with the v30 image (`expected/SFig14_v30.png`).
Then every number in `NUMBERS.tsv` is printed in table order.

```python
if FONT_DIR:
    from tools.check_pixels import compare
    for new, exp, stem in [("panel_abcd.png", "panel_abcd.png", "panel_abcd_png"),
                           ("panel_abcd.pdf", "panel_abcd.pdf", "panel_abcd_pdf"),
                           ("SFig14_v28f.png", "SFig14_v28f.png", "SFig14_v28f_png"),
                           ("SFig14_v28f.pdf", "SFig14_v28f.pdf", "SFig14_v28f_pdf"),
                           ("SFig14_v30.png", "SFig14_v35.png", "SFig14_v35")]:
        r = compare(OUT / new, UNIT / "expected" / exp, UNIT / "checks" / stem, dpi=300)
        print(new, "vs expected/" + exp + ":", r["verdict"], r.get("differing_pixels"))
else:
    print("pixel checks SKIPPED: fonts absent")

n = panels.stat_n_cases(SD)
print(n)
rows = [("465", "Four selected PREDICT spectra", n["word"])]
for line, quote, value in rows:
    print(line, "|", quote, "|", value)
```
