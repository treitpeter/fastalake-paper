# Supplementary Figure 13

Sequence context around histidine boundaries in CapScan, and the PREDICT replication. Panels a-f
show, for CapScan proximal fluid, distal fluid and stool and for N- and C-terminal boundaries, how
often each residue occurs at each flanking position of histidine-ending boundaries compared with
the source proteins (participant means, percentage points); dots mark q < 0.05 in the complete
32,400-test motif family. Panels g,h show PREDICT histidine enrichment as 93 patient means per
boundary side, for de novo predictions and for the non-specific search.

The notebook draws the producer figure (the v28f image) and the v30 house-style redraw of the
same values (`SFig13_v30`), which is the figure placed from v30 on.

Run from the repository root: `python paper.py build SFig13`.

```python
from pathlib import Path
import sys
UNIT = Path.cwd()                 # the notebook runs with its own directory as cwd
REPO = UNIT.parents[1]
sys.path[:0] = [str(REPO), str(UNIT)]
SD = REPO / "source_data" / UNIT.name
OUT = REPO / "build" / UNIT.name   # rendered panels; gitignored
OUT.mkdir(parents=True, exist_ok=True)
from style import fl_style as fl
import panels
fl.init()
```

## Panel a

CapScan proximal fluid (regions 1 and 2), N-terminal boundary: residue frequency at flanking
positions -4..+4 of boundaries whose P1 residue is histidine, observed minus source-protein
expected, participant means in percentage points. The conditioned position (-1) is omitted. Dots
mark q < 0.05 (Benjamini-Hochberg) in the complete 32,400-test motif family.

Claim: legend, "Dots mark q<0.05 in the full 32,400-test motif family, of which 840 cells are
displayed" (v28d line 461). n = 15 participants per cell. Test: two-sided exact sign-flip test of
the participant mean (all 2^15 sign assignments), BH within the family.

The figure is drawn once, as the producer drew it: all eight panels in one matplotlib figure, under
matplotlib defaults and the producer's settings (typeface DejaVu Sans), then the in-image title
is redacted and the page rendered at 300 dpi. Each panel section shows its crop of that render.
The display table is derived from the Figure 6 copy of the A28 CapScan family table and must be
byte-identical to this unit's Source Data table.

```python
import io
import pandas as pd
from IPython.display import Image, display


def show(png):
    """Show a rendered panel by relative link (not embedded: see OPEN_ITEMS.md, shared requests)."""
    display(Image(url=Path("../../build", UNIT.name, png.name).as_posix(), width=420))


FAM = REPO / "source_data" / panels.F_CAPSCAN_FAMILIES[0] / panels.F_CAPSCAN_FAMILIES[1]   # owned by Fig6
panels.check_sha256(FAM, panels.SHA256[panels.F_CAPSCAN_FAMILIES[1]])
for f in [panels.F_CAPSCAN_DISPLAY, panels.F_CAPSCAN_PARTICIPANT_MOTIF, panels.F_PREDICT_ACQUISITION, panels.F_PREDICT_FAMILIES]:
    panels.check_sha256(SD / f, panels.SHA256[f])
flanks, disp, disp_text = panels.derive_display_table(FAM)
buf = io.StringIO(); disp.to_csv(buf, sep="\t", index=False)
assert buf.getvalue() == (SD / panels.F_CAPSCAN_DISPLAY).read_text(), "display table differs from source_data"
print("display table rebuilt from the family table is identical to source_data:", True, "; cells:", len(disp))
allhist = panels.export_all_tests(FAM, OUT / "CapScan_histidine_all_tests.tsv")
print("all-tests export identical to the producer's Source Data export:",
      panels.sha256(OUT / "CapScan_histidine_all_tests.tsv") == panels.SHA256_CAPSCAN_ALL_TESTS_EXPORT)

pt = panels.predict_patient_means(SD / panels.F_PREDICT_ACQUISITION)
st = pd.read_csv(SD / panels.F_PREDICT_FAMILIES, sep="\t")
summary = panels.draw_figure(disp, pt, st, OUT / "HISTIDINE_CONTEXT_REPLICATION.pdf")
print("redaction:", panels.redact_and_rasterise(OUT / "HISTIDINE_CONTEXT_REPLICATION.pdf", OUT / "SFig13_v28f.pdf"))
crops = panels.crop_panels(OUT / "SFig13_v28f.png", summary["panel_boxes_in"], OUT, "panel_")
print("colour limit (pp):", summary["scale"])
print("panel a:", summary["a"])
show(crops["a"])
```

## Statistics

The 32,400-test CapScan motif family is recomputed from the participant-region motif differences,
and the 1,560-test PREDICT source family from the pseudonymised acquisition table, with the A28
sign-flip, bootstrap and Benjamini-Hochberg code. Each is compared row by row with the published
family table. The patient means are also compared with Figure 6's PREDICT patient table, which
uses the same pseudonyms. This takes about three minutes.

```python
import numpy as np
mf = panels.stat_motif_family(SD / panels.F_CAPSCAN_PARTICIPANT_MOTIF)
pubm = allhist[allhist.family.eq("all_motif_positions_residues")]
cm = panels.compare_family(mf, pubm, ["route", "side", "origin", "population", "position", "residue", "contrast"])
print("motif family, recomputed vs published:", cm)
assert cm["matched_rows"] == len(pubm) == len(mf) and cm["p"]["equal"] == len(mf)
assert all(cm[c]["max_abs_diff"] < 1e-12 for c in ["mean", "q", "CI95_low", "CI95_high"])

pf = panels.stat_predict_source_family(SD / panels.F_PREDICT_ACQUISITION)
pubp = st[st.family.eq("source_P1_all_residues")]
cp = panels.compare_family(pf, pubp, ["route", "side", "origin", "residue", "contrast"])
print("PREDICT source family, recomputed vs published:", cp)
assert cp["matched_rows"] == len(pubp) == len(pf)
assert all(cp[c]["max_abs_diff"] < 1e-12 for c in ["n", "mean", "p", "q", "CI95_low", "CI95_high"])

f6 = pd.read_csv(REPO / "source_data" / "Fig6" / "predict_histidine_patient_enrichment.tsv", sep="\t")
mm = f6.merge(pt.rename(columns={"participant": "patient"}), on=["route", "side", "origin", "residue", "patient"], suffixes=("_f6", ""))
assert len(mm) == len(f6) and np.allclose(mm.difference, mm.difference_f6, rtol=0, atol=1e-12)
print("patient means agree with Figure 6's table on", len(mm), "rows (same pseudonyms)")

counts = panels.stat_display_counts(disp, mf)
pos = panels.stat_predict_all_positive(pt)
print("display counts:", counts)
for k, v in pos.items():
    print("patients positive", k, v)
print("motif family: tests by participant n:", mf.n.value_counts().sort_index().to_dict(),
      "; estimable:", int(mf.estimable.sum()), "; q<0.05:", int((mf.q < panels.Q_CUTOFF).sum()))
print("PREDICT source family: estimable:", int(pf.estimable.sum()), "of", len(pf))
cols = ["route", "side", "n", "mean", "CI95_low", "CI95_high", "p", "null_hits", "q", "n_positive", "n_negative"]
shown = pf[pf.route.isin(panels.PREDICT_ROUTES) & pf.origin.eq("All") & pf.residue.eq("H")].copy()
# assignments (of 200,000) with |mean| at least the observed |mean|: p = (hits + 1) / 200,001
shown["null_hits"] = (shown.p * (panels.SIGNFLIP_DRAWS + 1)).round().astype(int) - 1
shown = shown[cols]
with pd.option_context("display.precision", 17, "display.width", 200):
    print(shown.to_string(index=False))
```

## Checks and numbers

Pixel checks against the original producer's outputs: the unredacted and the redacted PDF at
300 dpi, the 300 dpi composite PNG, and each panel cut from both composites with the same box. The
producer drew all text in DejaVu Sans, which ships with matplotlib, so these checks do not need the
Söhne fonts and always run. The v30 house-style composite is compared with the v30 image
(`expected/SFig13_v30.png`) only when `FASTALAKE_FONT_DIR` is set. Then every recomputed number, in `NUMBERS.tsv` order.

```python
import os
FONT_DIR = os.environ.get("FASTALAKE_FONT_DIR")
from tools.check_pixels import compare
E = UNIT / "expected"
pairs = [(OUT / "HISTIDINE_CONTEXT_REPLICATION.pdf", E / "HISTIDINE_CONTEXT_REPLICATION.pdf", "HISTIDINE_CONTEXT_REPLICATION"),
         (OUT / "SFig13_v28f.pdf", E / "SFig13_v28f.pdf", "SFig13_v28f_pdf"),
         (OUT / "SFig13_v28f.png", E / "SFig13_v28f.png", "SFig13_v28f")]
exp_crops = panels.crop_panels(E / "SFig13_v28f.png", summary["panel_boxes_in"], OUT, "expected_panel_")
pairs += [(crops[k], exp_crops[k], f"panel_{k}") for k in "abcdefgh"]
if FONT_DIR:   # the v30 house style is set in Söhne
    pairs.append((OUT / "SFig13_v30.png", E / "SFig13_v30.png", "SFig13_v30"))
else:
    print("v30 pixel check SKIPPED: fonts absent")
for new, exp, stem in pairs:
    r = compare(new, exp, UNIT / "checks" / stem, dpi=300)
    print(stem, r["verdict"], "differing pixels:", r.get("differing_pixels"))
print("composite PNG byte-identical to the producer's:", panels.sha256(OUT / "SFig13_v28f.png") == panels.sha256(E / "SFig13_v28f.png"))

def sole(values):
    assert len(values) == 1, values
    return next(iter(values))

numbers = [
    ("175", "patients positive, de novo top 30%, C boundary", pos[("denovo_top30", "C")]["n_positive"]),
    ("175", "patients positive, non-specific Sage both q, C boundary", pos[("sage_unspecific_bothq", "C")]["n_positive"]),
    ("461", "q cutoff for dots", panels.Q_CUTOFF),
    ("461", "motif family tests", counts["motif_family_tests"]),
    ("461", "displayed cells", counts["display_cells"]),
    ("461", "patient means per group (identical in all four g,h groups)", sole({v["n"] for v in pos.values()})),
    ("461", "bootstrap interval level (%)", round((0.975 - 0.025) * 100, 10)),
]
for line, what, v in numbers:
    print(line, what, repr(v), sep="\t")
```
