# Figure 2

Recovery of the defined CAMPI SIHUMIx community (15 acquisitions, eight members) by five search
databases: FastaLake (FL), the supplied reference proteomes (REF; "Reference" in the figure), the
multi-omic co-assembly proteins (MG), and FL with REF appended alone or with MG. The notebook
redraws the v30 producer canvas from `source_data/` (panels b and d-g at 9-50 residues, c and h at
5-50), applies the composite PDF edits (`assembly.py`), checks the composite pixel by pixel against
the v30 figure, and recomputes every number the v30 text and legend print.

Run from the repository root: `python paper.py build Fig2`.


```python
from pathlib import Path
import sys
UNIT = Path.cwd()                 # the notebook runs with its own directory as cwd
REPO = UNIT.parents[1]
sys.path[:0] = [str(REPO), str(UNIT)]
SD = REPO / "source_data" / UNIT.name
OUT = REPO / "build" / UNIT.name   # rendered panels; gitignored
OUT.mkdir(parents=True, exist_ok=True)
from style import fl_style as fl
import panels
fl.init()
```

## Checks and numbers

Composite: the canvas is turned into the v30 figure by the producer's PDF edits (`assembly.py`), and
the 300 dpi composite is compared pixel by pixel with the v30 figure (`expected/Fig2_v30.png`). The
v30 producer did not keep per-panel renders, so the composite is the reference. Text pixels match
only with the Söhne fonts (`FASTALAKE_FONT_DIR`); without them the composite and the check are
skipped and committed receipts are kept. Then the producer's independent recount of all 15
membership tables, and every recomputed number in `NUMBERS.tsv` order (`None` = OPEN).


```python
import os
from tools.check_pixels import compare
if os.environ.get("FASTALAKE_FONT_DIR"):
    import assembly
    comp = assembly.assemble(canvas, OUT)
    r = compare(OUT / "Figure2.png", UNIT / "expected" / "Fig2_v30.png", UNIT / "checks" / "composite", dpi=300)
    print("composite", r["verdict"], "differing pixels:", r.get("differing_pixels"))
else:
    print("pixel checks SKIPPED: fonts absent")

rc = panels.recount_membership(REPO / "source_data", d)
print("recount:", rc["checks"])
f = panels.stat_f_members(d, rc)
assert f["mg_zero_members_9"] == ["Bifidobacterium longum", "Lactobacillus plantarum"]
assert all(f["flr_exceeds_ref"].values()) and all(f["fl_ge_mg"].values())
print("least represented:", f["least_represented"], "FL >= MG:", f["fl_ge_mg"], "FL + REF >= REF:", f["flr_exceeds_ref"])
for line, name, value in panels.recompute_numbers(REPO / "source_data", d, rc):
    print(line, name, value, sep="\t")
```
